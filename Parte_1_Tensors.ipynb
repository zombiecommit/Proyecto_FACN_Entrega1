Tensores en PyTorch

Introducción

Los tensores son estructuras de datos que permiten almacenar y manipular valores numéricos en una o varias dimensiones. En PyTorch se utilizan para realizar operaciones matemáticas y constituyen la base del procesamiento de datos en las redes neuronales.

En este notebook se exploran la creación de tensores, las operaciones aritméticas, la selección de elementos, la modificación de dimensiones y la interoperabilidad entre PyTorch y NumPy.

```python
x = torch.rand(3, 2)
x
```

```python
y = torch.ones(x.size())
y
```

```python
z = x + y
z
```

Creación y operaciones matemáticas

torch.rand(3, 2) crea un tensor de 3 filas y 2 columnas con valores aleatorios. Por su parte, torch.ones(x.size()) crea otro tensor con las mismas dimensiones que x, pero con todos sus elementos iguales a uno.

La operación x + y suma los elementos correspondientes de ambos tensores. Para que esta operación funcione, sus dimensiones deben ser compatibles.

El resultado permite observar cómo PyTorch realiza operaciones matemáticas directamente sobre estructuras multidimensionales.

Indexación y selección de elementos

La indexación permite acceder a elementos específicos de un tensor. En Python, los índices comienzan en cero.

- z[0] selecciona la primera fila.
- z[0, 1] selecciona el elemento de la primera fila y la segunda columna.
- z[:, 0] selecciona todas las filas de la primera columna.

El operador : permite seleccionar todos los elementos de una dimensión. Estas operaciones facilitan la consulta y manipulación de partes específicas de los datos.

```python
z[0]
```

```python
z[0, 1]
```

```python
z[:, 0]
```

Operaciones in-place y dimensiones

Las operaciones *in-place* modifican directamente el tensor sobre el que se ejecutan. Por ejemplo, z.add_(x) suma x a z y guarda el resultado en el mismo tensor.

En cambio, z.add(1) calcula un resultado nuevo sin modificar directamente el tensor original.

Las dimensiones pueden consultarse con z.size(). Además, resize_() permite modificar el tamaño del tensor, mientras que reshape() permite reorganizar sus elementos en una forma diferente cuando el número de elementos es compatible.

Es importante distinguir estas operaciones porque algunas modifican el tensor original y otras devuelven un resultado nuevo.

```python
z.add_(x)
z
```

```python
z.size()
```

```python
z.resize_(1, 6)
z
```

Interoperabilidad entre NumPy y PyTorch

NumPy permite trabajar con arreglos multidimensionales, mientras que PyTorch ofrece tensores con operaciones diseñadas para el aprendizaje profundo.

torch.from_numpy(a) crea un tensor a partir de un arreglo de NumPy. En este caso, ambos comparten la misma memoria, por lo que modificar los valores del arreglo también puede reflejarse en el tensor.

Por otro lado, b.numpy() permite obtener un arreglo de NumPy a partir de un tensor compatible. Esta interoperabilidad facilita combinar herramientas de análisis numérico con operaciones de PyTorch.

```python
a = np.random.rand(3, 2)
a
```

```python
b = torch.from_numpy(a)
b
```

```python
a[0, 0] = 10
a
```

```python
b
```

```python
c = b.numpy()
c
```
